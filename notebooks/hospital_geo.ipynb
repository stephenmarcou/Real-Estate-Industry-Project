```python
import pandas as pd
import geopandas as gpd
import folium
```

```python
geoJSON = gpd.read_file("../data/curated/geojson/shapefile.geojson")
hospital_data = pd.read_csv("../data/curated/hospital_data/hospital_data_curated.csv")
```

```python
# put this in preprocessing
print(len(hospital_data))
hospital_data = hospital_data[~hospital_data["coordinate"].isnull()]
print(len(hospital_data))
```

```python
hospital_data["coordinate"] = hospital_data["coordinate"].apply(eval)
```

```python
_map = folium.Map(location=[-37.840935, 144.9631], tiles="Stamen Terrain", zoom_start=10)

_map.add_child(folium.Choropleth(
    geo_data=geoJSON,
    name='choropleth'
))
```

```python
for hospital_name, coord in hospital_data[["formal name", "coordinate"]].values:
    _map.add_child(
        folium.Marker(location=coord, popup=hospital_name)
    )
```
