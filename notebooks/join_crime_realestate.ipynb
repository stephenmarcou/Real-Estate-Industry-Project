```python
import pandas as pd
```

```python
rental_data_sf = pd.read_csv("../data/curated/rental_data_sf.csv")
```

```python
rental_data_sf.head()
```

```python
crime_data =  pd.read_csv("../data/raw/crime_data.csv")
```

```python
crime_data_23 = crime_data[crime_data['Year'] == 2023]
crime_data_23_agg = crime_data_23.groupby(['Postcode']).agg({'Offence Count': 'sum'})
crime_data_23_agg.reset_index()

crime_data_22 = crime_data[crime_data['Year'] == 2022]
crime_data_22_agg = crime_data_22.groupby(['Postcode']).agg({'Offence Count': 'sum'})
crime_data_22_agg.reset_index()

crime_data_21 = crime_data[crime_data['Year'] == 2022]
crime_data_21_agg = crime_data_21.groupby(['Postcode']).agg({'Offence Count': 'sum'})
crime_data_21_agg.reset_index()
```

```python
rental_data_crime_join = pd.merge(rental_data_sf, crime_data_23_agg, on='Postcode', how='left')
rental_data_crime_join = rental_data_crime_join.rename(columns={'Offence Count': 'Offence Count 2023'})

rental_data_crime_join = pd.merge(rental_data_crime_join, crime_data_22_agg, on='Postcode', how='left')
rental_data_crime_join = rental_data_crime_join.rename(columns={'Offence Count': 'Offence Count 2022'})

rental_data_crime_join = pd.merge(rental_data_crime_join, crime_data_21_agg, on='Postcode', how='left')
rental_data_crime_join = rental_data_crime_join.rename(columns={'Offence Count': 'Offence Count 2021'})

rental_data_crime_join.head()
```

```python
folder_path = '../data/curated/'
file_name = 'rental_data_crime_join.csv'

rental_data_crime_join.to_csv(folder_path + file_name, index=False)
```
