```python
import geopandas as gpd
import pandas as pd
```

```python
open_spaces_sf = gpd.read_file("../data/raw/open_space/VPA_Draft_Open_Space_Data.shp")
sa2_districts_sf = gpd.read_file("../data/landing/sa2/SA2_2021_AUST_GDA2020.shp")

open_spaces_sf.head()
```

```python
joined = gpd.sjoin(open_spaces_sf, sa2_districts_sf[['SA2_CODE21', 'SA2_NAME21', 'geometry']], how="left", op="within")

joined.head()
```

```python
# create a JSON 
geoJSON = open_spaces_sf[['FID', 'geometry']].drop_duplicates('FID').to_json()

# print the first 300 chars of the json
print(geoJSON[:300])
```

```python
import folium
```

```python
_map = folium.Map(location=[-37.020100, 144.964600], tiles="Stamen Terrain", zoom_start=10)

# refer to the folium documentations on how to plot aggregated data.
_map.add_child(folium.Choropleth(
    geo_data=geoJSON,
    name='choropleth',
))

_map
```
