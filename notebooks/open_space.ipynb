```python
rent_data = pd.read_csv("../data/curated/rental_data.csv")

rent_data.head()
```

Convert the coordinates to GeoPandas Geometry type in shape files

```python
from shapely.geometry import Point
import ast

# Convert the string to a list of floats
rent_data['coordinates'] = rent_data['coordinates'].apply(ast.literal_eval)

# Convert the list of floats to Point geometry (note the order: (longitude, latitude))
rent_data['geometry'] = rent_data['coordinates'].apply(lambda x: Point(x[1], x[0]))


# Save to shapefile
# grent_data.to_file("output_points.shp")
```

```python
# Convert DataFrame to GeoDataFrame
rent_data_sf = gpd.GeoDataFrame(rent_data, geometry='geometry', crs="EPSG:4326")

rent_data_sf.head()
```

```python
rent_data_sf = gpd.sjoin(rent_data_sf, sa2_districts_sf[['SA2_CODE21', 'SA2_NAME21', 'geometry']], how="left", op="within")

rent_data_sf.head()
```

```python
parks.shape
```

```python
# 1. Group parks by SA2 district and count
park_counts = parks.groupby('SA2_CODE21').size().reset_index(name='num_parks')

# 2. Merge park_counts with rent_data on the SA2 district column
result = pd.merge(rent_data_sf, park_counts, on='SA2_CODE21', how='left')

# Fill NaNs with 0 - in case there are properties with no parks in their SA2 district
result['num_parks'].fillna(0, inplace=True)

print(result[['SA2_CODE21', 'num_parks']])
```

```python
# parks['park_area'] = parks['geometry'].area

# 2. Group parks by SA2 district and sum the areas
park_area_sum = parks.groupby('SA2_CODE21')['HA'].sum().reset_index()

# 3. Merge park_area_sum with rent_data on the SA2 district column
result = rent_data_sf.merge(park_area_sum, on='SA2_CODE21', how='left')


# # Fill NaNs with 0 - in case there are properties in SA2 districts with no parks
result['HA'].fillna(0, inplace=True)

print(result[['SA2_CODE21', 'HA']])
```

```python
# Buffer the melb_rent by 5km. Note: Ensure your data is in a CRS that uses meters for distance!
# If not, you might need to reproject your GeoDataFrames to a suitable CRS.
melb_rent_buffered = melb_rent.copy()
melb_rent_buffered['geometry'] = melb_rent_buffered.buffer(5000) 


# Drop 'index_left' and 'index_right' columns if they exist in melb_rent_buffered
if 'index_left' in melb_rent_buffered.columns:
    melb_rent_buffered = melb_rent_buffered.drop(columns='index_left')

if 'index_right' in melb_rent_buffered.columns:
    melb_rent_buffered = melb_rent_buffered.drop(columns='index_right')

# Drop 'index_left' and 'index_right' columns if they exist in parks
if 'index_left' in parks.columns:
    parks = parks.drop(columns='index_left')

if 'index_right' in parks.columns:
    parks = parks.drop(columns='index_right')
    
    
# Perform a spatial join between the buffered melb_rent and parks
test_sdf = gpd.sjoin(melb_rent_buffered, parks, how='left', op='contains')

# Count the number of parks for each entry in melb_rent
parks_count = test_sdf.groupby('index_left').size()

# Add the count back to the melb_rent GeoDataFrame if desired
melb_rent['parks_count'] = parks_count
```

```python
# Function to return the index of the nearest open space
def nearest_open_space(point, open_spaces_df):
    distances = open_spaces_df.distance(point)
    return distances.idxmin()

# For each point, find the nearest open space
rent_data_sf['nearest_open_space'] = rent_data_sf['geometry'].apply(lambda point: nearest_open_space(point, joined))
```
