```python
import geopandas as gpd

sa2_sf = gpd.read_file("../data/landing/sa2/SA2_2021_AUST_GDA2020.shp")
```

```python
sa2_sf = sa2_sf.loc[sa2_sf["STE_NAME21"] == "Victoria"]
```

Change all SA2 codes such that it starts with a 2

```python
vic_postcodes = vic_postcodes[vic_postcodes['SA2_MAINCODE_2016'].astype(str).str.startswith('2')]
```

```python
from shapely.geometry import Point
import ast

vic_postcodes['geometry'] = [Point(xy) for xy in zip(vic_postcodes['long'], vic_postcodes['lat'])]
```

```python
vic_postcodes.head()
```

```python
vic_postcodes_restricted = vic_postcodes[["id", "postcode", "locality", "state", "long", "lat", "geometry"]]
```

```python
vic_postcodes_sf = gpd.GeoDataFrame(vic_postcodes_restricted, geometry='geometry', crs="EPSG:4326")
```

```python
vic_postcodes_sf = gpd.sjoin(vic_postcodes_sf, sa2_sf[['SA2_CODE21', 'SA2_NAME21', 'geometry']], how="left", op="within")

```

```python
vic_postcodes_sf = vic_postcodes_sf.drop(columns='index_right')
vic_postcodes_sf.head()
```
