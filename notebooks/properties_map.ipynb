```python
import pandas as pd
import geopandas as gpd
import numpy as np
import folium
```

```python
property_data = pd.read_csv("../data/curated/rental_data_sf.csv")
property_data.columns = property_data.columns.str.lower()
# count number of properties at each sa2 area
property_sa2_count = property_data.groupby(by=["sa2_code21"]).size().reset_index(name="count")
property_sa2_count["log count"] = np.log(property_sa2_count["count"] + 1)
property_sa2_count
```

```python
geoJSON = gpd.read_file("../data/curated/geojson/shapefile.geojson")
```

```python
geoJSON.keys()
```

```python
print(geoJSON[:300])
```

```python
m = folium.Map(location=[-37.020100, 144.964600], tiles="Stamen Terrain", zoom_start=7)

#geoJSON = gdf[['LGA_CODE23', 'geometry']].drop_duplicates('LGA_CODE23').to_json()

# refer to the folium documentations on more information on how to plot aggregated data.
c = folium.Choropleth(
    geo_data=geoJSON, # geoJSON 
    name='choropleth', # name of plot
    data=property_sa2_count.reset_index(), # data source
    columns=['sa2_code21','log count'], # the columns required
    key_on="properties.SA2_CODE21",
    fill_color='YlOrRd', # color scheme
    nan_fill_color='black',
    legend_name='Number of rental properties available'
    
)

c.add_to(m)

```
