Find the SA2 district for each rental property

```python
import geopandas as gpd
import pandas as pd
```

```python
sa2_districts_sf = gpd.read_file("../data/landing/sa2/SA2_2021_AUST_GDA2020.shp")
rent_data = pd.read_csv("../data/curated/rental_data.csv")
```

```python
from shapely.geometry import Point
import ast

# Convert the string to a list of floats
rent_data['coordinates'] = rent_data['coordinates'].apply(ast.literal_eval)

# Convert the list of floats to Point geometry (note the order: (longitude, latitude))
rent_data['geometry'] = rent_data['coordinates'].apply(lambda x: Point(x[1], x[0]))
```

```python
# Convert DataFrame to GeoDataFrame
rent_data_sf = gpd.GeoDataFrame(rent_data, geometry='geometry', crs="EPSG:4326")

rent_data_sf.head()
```

```python
rent_data_sf = gpd.sjoin(rent_data_sf, sa2_districts_sf[['SA2_CODE21', 'SA2_NAME21', 'geometry']], how="left", op="within")

rent_data_sf.head()
```

```python
rent_data_sf = rent_data_sf.drop(columns='index_right')
```

```python
rent_data_sf.to_csv("../data/curated/rental_data_sf.csv", index=False)
```

```python
rent_data_sf = pd.read_csv("../data/curated/rental_data_sf.csv")
```

```python
from shapely import wkt
```

```python
rent_data_sf['geometry'] = rent_data_sf['geometry'].apply(wkt.loads)
```

```python
rent_data_sf = rent_data_sf[rent_data_sf['SA2_CODE21'].notnull()]

rent_data_sf = rent_data_sf[rent_data_sf['Postcode'].notnull()]

rent_data_sf.SA2_CODE21 = rent_data_sf.SA2_CODE21.astype(int)

rent_data_sf.Postcode = rent_data_sf.Postcode.astype(int)
```

```python
rent_data_sf.head()
```

```python
rent_data_sf[["Postcode", "SA2_CODE21"]]
```

```python
postcode_sa2_conversion = rent_data_sf[['Postcode', 'SA2_CODE21']].drop_duplicates(subset=['Postcode', 'SA2_CODE21'])
```

```python
postcode_sa2_conversion
```

```python
postcode_sa2_conversion.groupby("Postcode").size()
```

```python
sa2_info = pd.read_csv("../data/curated/suburb_info.csv")

sa2_info.head()
```

```python
# Merge the dataframes on the relevant columns
merged_df = pd.merge(postcode_sa2_conversion, sa2_info, left_on='SA2_CODE21', right_on='SA2_codes', how='inner')

# Perform groupby operation and aggregate the columns as specified
agg_funcs = {
    'Estimated resident population (no.)': 'sum',
    'Employed (no.)': 'sum',
    'Unemployed (no.)': 'sum',
    'Land area (ha)': 'sum',
    'Total families (no.)': 'sum',
    'Total dwellings (no.)': 'sum',
    'Median equivalised total household income (weekly) ($)': 'mean',
    'Median weekly household rental payment ($)': 'mean'
}

result_df = merged_df.groupby('Postcode').agg(agg_funcs).reset_index()

print(result_df)
```

```python
merged_df.to_csv("../data/curated/postcode_info_2021.csv", index=False)
```
