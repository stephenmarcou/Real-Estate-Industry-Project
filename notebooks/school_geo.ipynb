```python
import pandas as pd
import geopandas as gpd
import folium
```

```python
school_data = pd.read_csv("../data/curated/school_data/school_data.csv")
geoJSON = gpd.read_file("../data/curated/geojson/shapefile.geojson")
# ensure coordinate is tuple type
school_data["coordinate"] = school_data["coordinate"].apply(eval)
school_data
```

```python
top_secondary_schools = school_data[school_data["median vce score"].notnull()]
```

```python
school_data
```

```python
top_secondary_schools
```

```python

_map = folium.Map(location=[-37.840935, 144.9631], tiles="Stamen Terrain", zoom_start=10)

_map.add_child(folium.Choropleth(
    geo_data=geoJSON,
    name='choropleth'
))

```

```python
for school_name, coord in top_secondary_schools[["school_name", "coordinate"]].values:
    _map.add_child(
        folium.Marker(location=coord, popup=school_name)
    )
```
