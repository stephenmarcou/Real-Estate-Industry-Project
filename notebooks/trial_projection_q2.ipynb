```python
cols = ["Estimated resident population (no.)", "Population density (persons/km2)",
        "Employed (no.)", "Unemployed (no.)", "In the Labour force (no.)", "Total persons aged 15 years and over (no.)",
        'Median equivalised total household income (weekly) ($)',
        'Land area (ha)', 'Median weekly household rental payment ($)', 
        'Average family size (no. of persons)', 'Total families (no.)', 'Total dwellings (no.)']

sa2_codes = suburb_data_2011["SA2_codes"].values
```

```python
def compute_growth_rate(df, col_name):
    """
    Calculate the average growth rate of a specific column for a given DataFrame.

    This function calculates the average growth rate for a specific column 'col_name' in the provided DataFrame 'df' by
    comparing values between three consecutive rows. The growth rate is computed as the percentage change between the
    second and first row and the percentage change between the first and zeroth row, then averaging these two growth rates.

    Parameters:
    df (DataFrame): A pandas DataFrame containing the data for which the growth rate is to be calculated.

    Returns:
    float: The average growth rate of the specified column.
    """
    growth_rate_1 = (df.iloc[2][col_name] - df.iloc[1][col_name]) / df.iloc[1][col_name]
    growth_rate_2 = (df.iloc[1][col_name] - df.iloc[0][col_name]) / df.iloc[0][col_name]
    return (growth_rate_1 + growth_rate_2)/2
```

```python
suburb_data_2011['Year'] = 2011
suburb_data_2016['Year'] = 2016
suburb_data_2021['Year'] = 2021

combined_df = pd.concat([suburb_data_2011, suburb_data_2016, suburb_data_2021])
```

```python
growth_rate = (pd.DataFrame(columns= cols, index=sa2_codes)).rename_axis('SA2_Code', axis=1)
```

Compute the Growth rate for each column

```python
for sa2_code in sa2_codes:
    for col in cols:
        df_postcode = combined_df[combined_df['SA2_codes'] == sa2_code].sort_values(by="Year")
        growth_rate.loc[sa2_code][col] = compute_growth_rate(df_postcode, col)
```

### Apply the growth rates to the 2021 census data

```python
growth_rate = pd.read_csv("../data/curated/growth_rates.csv")
```

```python
growth_rate.head()
```

```python
suburb_data_2021 = pd.read_csv("../data/curated/suburb_info_2021.csv")

suburb_data_2021 = suburb_data_2021.set_index("SA2_codes")
```

```python
growth_rate.head()
```

```python
growth_rate = growth_rate.set_index("SA2_codes")
```

```python
result = suburb_data_2021.multiply(growth_rate + 1)
```

```python
result.head()
```

Now we just need to round the values

```python
# List of columns to exclude from rounding
exclude_cols = ['Median equivalised total household income (weekly) ($)', 
                'Median weekly household rental payment ($)', 
                'Average family size (no. of persons)']

# Columns to be rounded
cols_to_round = [col for col in result.columns if col not in exclude_cols]

# Round the selected columns
result[cols_to_round] = result[cols_to_round].round()

```
