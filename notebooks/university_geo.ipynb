```python
import pandas as pd
import geopandas as gpd
import folium
```

```python
# read sa2 shapefile and university data
geoJSON = gpd.read_file("../data/curated/geojson/shapefile.geojson")
university_data = pd.read_csv("../data/curated/school_data/university_location_curated.csv")
```

```python
university_data["coordinate"] = university_data["coordinate"].apply(eval)
```

```python
_map = folium.Map(location=[-37.840935, 144.9631], tiles="Stamen Terrain", zoom_start=10)

_map.add_child(folium.Choropleth(
    geo_data=geoJSON,
    name='choropleth'
))
```

```python
for university_name, coord in university_data[["institution", "coordinate"]].values:
    _map.add_child(
        folium.Marker(location=coord, popup=university_name)
    )
```
